```python
# ============================================
# Smart Credit Card Processing System (Advanced)
# ============================================

import random

# Generate Transaction ID
def generate_transaction_id():
    return "TXN" + str(random.randint(10000, 99999))

# Fraud Detection Logic
def detect_fraud(amount):
    if amount > 15000:
        return "HIGH RISK"
    elif amount > 8000:
        return "MEDIUM RISK"
    else:
        return "LOW RISK"

# Process Payment
def process_payment(amount):
    if amount <= 0:
        return False, "Invalid Amount"
    return True, "Payment Initiated"

# Main System
def credit_card_system():
    print("====================================")
    print("💳 Smart Credit Card System Started")
    print("====================================")

    while True:
        print("\n1. Make Payment")
        print("2. Exit")

        choice = input("Enter choice: ")

        if choice == '1':
            amount = float(input("Enter Amount: "))

            txn_id = generate_transaction_id()

            status, message = process_payment(amount)
            print("\nTransaction ID:", txn_id)
            print("Status:", message)

            if not status:
                continue

            risk = detect_fraud(amount)
            print("Risk Level:", risk)

            if risk == "HIGH RISK":
                print("🚫 Transaction Blocked (Fraud Detected)")
            elif risk == "MEDIUM RISK":
                print("⚠️ Transaction Under Review")
            else:
                print("✅ Payment Successful")

        elif choice == '2':
            print("Exiting System...")
            break

        else:
            print("Invalid Choice")

# Run System
credit_card_system()
```
